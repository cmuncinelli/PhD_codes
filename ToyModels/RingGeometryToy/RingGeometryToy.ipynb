````python
%load_ext autoreload
%autoreload 2

from IPython.display import Image, display

from ringtoy import explorations
from ringtoy.notebook_ui import RingExplorer
from ringtoy.scenarios import get_preset

# None writes to the default set in ringtoy/output.py.
OUTPUT_ROOT = None
````

## Interactive explorer

Tabs hold every parameter: jet direction, ring shape, placement (along the jet or fixed in the lab),
elliptic flow and almond, polarization and window, display. Play buttons animate the jet azimuth,
the distance along the jet, and the flow strength. Clicking a curve in the $\Delta\phi$ plot, or the
*highlight bin* slider, marks the ring elements that fill that bin in both 3D panels.
*export scene HTML* and *export movie of sweep* write standalone files with provenance.

````python
explorer = RingExplorer(output_root=OUTPUT_ROOT)
explorer.display()
````

Any preset can be the starting point, for example the jet--flow misalignment setup:

````python
misaligned_explorer = RingExplorer(get_preset("misaligned"), sweep_parameter="jet_phi", output_root=OUTPUT_ROOT)
misaligned_explorer.display()
````

## Inward/outward morph

The ring moves along the jet from $L=-4$ fm (inward) to $L=+4$ fm (outward). Expected: the cone on the
momentum sphere flips from $-\hat t$ to $+\hat t$, and the $\Delta\phi$ peaks migrate from $\pm\pi$ towards 0.

````python
fig, run_dir = explorations.inward_outward_morph(output_root=OUTPUT_ROOT)
print(run_dir)
fig.show()
````

## Misalignment movie

The ring centre stays at $x=2$ fm while the jet azimuth turns once around; the side plot shows the
integrated response versus $\phi_{\rm Jet}$ with the current value marked.

````python
fig, run_dir = explorations.misalignment_movie(output_root=OUTPUT_ROOT)
print(run_dir)
fig.show()
````

## Jet pseudorapidity movie

The ring cone crosses the $|\eta|<0.5$ window as $\eta_{\rm Jet}$ runs from $-1.5$ to $1.5$.

````python
fig, run_dir = explorations.eta_jet_movie(output_root=OUTPUT_ROOT)
print(run_dir)
fig.show()
````

## Evolution movie

Ring and almond evolve with the same shape map. The almond starts elongated out of plane, becomes round,
and ends elongated in plane (eccentricity inversion); its eccentricity is quoted in the title.

````python
fig, run_dir = explorations.evolution_movie(output_root=OUTPUT_ROOT)
print(run_dir)
fig.show()
````

## Does the response depend on the jet angle to the event plane?

Relative variation $(\max-\min)/|{\rm mean}|$ of the integrated ring average versus
$\delta=\phi_{\rm Jet}-\Psi_2$. Small values mean that selecting jets relative to $\Psi_2$ is not worth it.

````python
variation, run_dir = explorations.delta_test(output_root=OUTPUT_ROOT)
print(variation)
display(Image(filename=f"{run_dir}/delta_test.png"))
````

## Ellipticity check

Toy ring average of a coaxial ellipse against the closed form $(b/a)\,K(k)/E(k)$.

````python
max_difference, run_dir = explorations.ellipticity_check(output_root=OUTPUT_ROOT)
print(f"max |toy - analytic| = {max_difference:.2e}")
display(Image(filename=f"{run_dir}/ellipticity_check.png"))
````
